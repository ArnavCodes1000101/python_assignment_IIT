```python
products = { "Milk": {"price in rupees": 60, "quantity": 10},
"Bread": {"price in rupees": 40, "quantity": 15},
"Eggs": {"price in rupees": 120, "quantity": 20},
"Rice": {"price in rupees": 80, "quantity": 25},
"Biscuits": {"price in rupees": 30, "quantity": 30} }

import pandas as pd
df = pd.DataFrame(products).T
print("Available Products:")
display(df)

cart = []
total = 0

while True:
    my_product = input("Enter the product you want to buy (q to quit): ")

    if my_product == 'q':
        break

    if my_product in products:
        try:
            quantity = int(input(f"Enter quantity for {my_product}: "))

            available_stock = products[my_product]["quantity"]
            if quantity <= available_stock:
                unit_price = products[my_product]['price in rupees']
                item_price = quantity * unit_price

                cart.append({
                    'product': my_product,
                    'quantity': quantity,
                    'unit_price': unit_price,
                    'total_price': item_price })

                total += item_price

                products[my_product]['quantity'] -= quantity
                print(f"{quantity} x {my_product} added to cart.")
            else:
                print(f"Insufficient stock for {my_product}. Available: {available_stock}.enter a lower quantity.")

        except ValueError:
            print("Invalid quantity. Please enter a valid number.")
    else:
        print(f"Product '{my_product}' not found. choose from  available products.")

print(" Your Shopping Cart")
for item in cart:
        print(f"- {item['product']} (x{item['quantity']}): ₹{item['unit_price']} each, Total: ₹{item['total_price']}")



        print(f"Grand Total of your purchase: ₹{total}")

display(pd.DataFrame(products).T)
```
