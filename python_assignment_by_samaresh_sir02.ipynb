```python
#question 4
a = int(input("Enter first side: "))
b = int(input("Enter second side: "))
c = int(input("Enter third side: "))

if a + b <= c or a + c <= b or b + c <= a:
    print("Invalid triangle")

elif a == b and b == c:
    print("Equilateral triangle")

elif a == b or b == c or a == c:
    print("Isosceles triangle")

else:
    print("Scalene triangle")
```

```python
#question 6
marks = [78, 45, 91, 33, 67, 49]

passed = 0
failed = 0

for mark in marks:

    if mark >= 40:
        print(mark, "Pass")
        passed = passed + 1

    else:
        print(mark, "Fail")
        failed = failed + 1

print("Failed students:", failed)
print("Passed students:", passed)

print("Fail : Pass =", failed, ":", passed)
```

```python
#question 7
stock = {
    "laptop": 5,
    "mouse": 0,
    "keyboard": 3,
    "monitor": 0
}

product = input("Enter product name: ")

if product not in stock:
    print("Product is not present")

elif stock[product] == 0:
    print("Product is unavailable")

else:
    print("Product is available")
    print("Quantity:", stock[product])
```

```python
#question 8
balance = int(input("Enter account balance: "))
withdraw = int(input("Enter amount to withdraw: "))

if withdraw <= 0:
    print("Invalid withdrawal amount")

elif withdraw > balance:
    print("Insufficient balance")

else:
    balance = balance - withdraw
    print("Withdrawal successful")
    print("Remaining balance:", balance)
```

```python
#question 9

shopping_amount = int(input("Enter shopping amount: "))

if shopping_amount <= 1000:
    discount = shopping_amount * 0

elif shopping_amount <= 4999:
    discount = shopping_amount * 0.1

elif shopping_amount <= 9999:
    discount = shopping_amount * 0.2

else:
    discount = shopping_amount * 0.3
print("actual_amount: ", shopping_amount)
print("Discount: ", discount)
print("Total amount to be paid: ", shopping_amount - discount)
```

```python
#question 10

products = {
"laptop": {"price": 50000, "stock": 3},
"phone": {"price": 30000, "stock": 5},
"headphones": {"price": 2000, "stock": 0}
}
print(products)
product = input("Enter product name: ")

if product not in products:
    print("Product not found")

elif products[product]["stock"] == 0:
    print("Product is out of stock")

else:
    print("Product is available")

quantity = int(input("Enter quantity: "))

if quantity > products[product]["stock"]:
    print("Insufficient stock")

else:
    total_price = products[product]["price"] * quantity
    print("Total price:", total_price)

if total_price >= 50000:
    discount = total_price * 0.1
    print("Discount:", discount)

final_price = total_price - discount

print("Final price:", final_price)
```

```python
#question 11
students = { "Aman": [85, 92, 78], "Riya": [95, 88, 91], "Karan": [72, 65, 80], "Simran": [45, 92, 88], "Raj": [90, 91, 95] }
scholarship = 0
highest_average = 0
highest_scorer = ""

for name, marks in students.items():
    average = sum(marks) / len(marks)

    category = ""

    if any(mark <= 40 for mark in marks):
        category = "Failed"
    else:

        if average >= 90 and all(mark >= 85 for mark in marks):
            category = "Scholarship"
            scholarship += 1
        elif average >= 80:
            category = "Distinction"
        elif average >= 60:
            category = "Pass"
        else:
            category = "Needs Improvement"

    if average > highest_average:
        highest_average = average
        highest_scorer = name

    print(name, ":", category)

print("Total Scholarship Students:", scholarship)
print("Highest Average:", highest_average)
print("Highest Scorer:", highest_scorer)
```

```python
#bonus question
transactions = {
"Aman": [500, 1200, 300, 15000, 700],
"Riya": [200, 300, 450, 600],
"Karan": [10000, 50, 80, 12000],
"Simran": [400, 500, 600, 700]
}
print(transactions)
highest_total = 0
highest_spending_customer = ""
total_suspicious = 0

for name, amounts in transactions.items():
    total = sum(amounts)
    suspicious = 0
    for amount in amounts:
        if amount > 10000:
            suspicious = suspicious + 1

    if suspicious >= 2 or (suspicious == 1 and total > 20000):
        risk = "High Risk"
    else:
        risk = "Low Risk"

    print(name, ": ", risk)
    print("Total Spent: ", total)
    print("Suspicious Transactions: ", suspicious)
    print()

    total_suspicious = total_suspicious + suspicious
    if total > highest_total:
        highest_total = total
        highest_spending_customer = name

print("Highest Spending Customer:", highest_spending_customer)
print("Highest Total Spending:", highest_total)
print("Total Suspicious Transactions in the data: ", total_suspicious)
```
