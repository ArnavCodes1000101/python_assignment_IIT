```python
import random

print("hello, lets guess my room number(3055-3075): ")

secret_room_number = random.randint(3070,3070)
max_tries = 5
tries_left = max_tries
flag = False

while not flag and tries_left > 0:

     user_guess = int(input("enter your guess: "))

     if user_guess == secret_room_number:
            print("Congratulations! You are right!")
            flag = True

     else:
            print("wrong guess try again")
            tries_left -= 1


if not flag:
    print("Sorry, you ran out of tries. The secret room number was 3070.")
```
