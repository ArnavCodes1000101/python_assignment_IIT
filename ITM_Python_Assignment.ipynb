```python
import numpy as np

n = int(input("how many values of x and y do you need?: "))

x = np.zeros(n)
y = np.zeros(n)

for i in range(n):
    x[i] = float(input("enter x" + str(i) + " value: "))
    y[i] = float(input("enter y" + str(i) + " value: "))


a = (n*np.sum(x*y) - np.sum(x)*np.sum(y))/ (n*np.sum(x**2) - np.sum(x)**2)

b = (np.sum(y) - a*np.sum(x))/n

original_equation= a*x + b

error = y - original_equation

E = np.sum(error**2)

print("a = ", a)
print("b = ", b)
print("E = ", E)

import matplotlib.pyplot as plt

plt.scatter(x, y)
plt.plot(x, original_equation)
plt.show()
```
