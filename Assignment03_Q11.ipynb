```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files

# Upload image
print("Upload a JPG or PNG image")
uploaded = files.upload()

filename = list(uploaded.keys())[0]
original = Image.open(filename).convert("RGB")

image = original.copy()


def show_image(img, title):
    plt.figure(figsize=(7, 5))
    plt.imshow(img)
    plt.title(title)
    plt.axis("off")
    plt.show()


while True:

    print("\n====== IMAGE TRANSFORMATION TOOLBOX ======")
    print("1. Rotate")
    print("2. Resize")
    print("3. Flip")
    print("4. Shear")
    print("5. Custom Matrix")
    print("6. Reset")
    print("7. Exit")

    choice = input("Enter your choice: ")

    # Rotate
    if choice == "1":
        angle = float(input("Enter rotation angle: "))

        image = image.rotate(
            angle,
            expand=True
        )

        show_image(image, "Rotated Image")


    # Resize
    elif choice == "2":
        factor = float(input("Enter resize factor: "))

        w, h = image.size

        new_w = int(w * factor)
        new_h = int(h * factor)

        image = image.resize((new_w, new_h))

        show_image(image, "Resized Image")


    # Flip
    elif choice == "3":
        print("1. Horizontal")
        print("2. Vertical")

        flip_choice = input("Choose flip direction: ")

        if flip_choice == "1":
            image = image.transpose(Image.Transpose.FLIP_LEFT_RIGHT)
        elif flip_choice == "2":
            image = image.transpose(Image.Transpose.FLIP_TOP_BOTTOM)
        else:
            print("Invalid choice")
            continue

        show_image(image, "Flipped Image")


    # Shear
    elif choice == "4":
        print("1. Horizontal shear")
        print("2. Vertical shear")

        shear_choice = input("Choose shear direction: ")
        shear = float(input("Enter shear value: "))

        w, h = image.size

        if shear_choice == "1":
            matrix = (1, shear, 0,
                      0, 1, 0)

        elif shear_choice == "2":
            matrix = (1, 0, 0,
                      shear, 1, 0)

        else:
            print("Invalid choice")
            continue

        image = image.transform(
            (w, h),
            Image.Transform.AFFINE,
            matrix
        )

        show_image(image, "Sheared Image")


    # Custom Matrix
    elif choice == "5":
        print("Enter a 2 x 2 transformation matrix")

        a = float(input("a = "))
        b = float(input("b = "))
        c = float(input("c = "))
        d = float(input("d = "))

        A = np.array([
            [a, b],
            [c, d]
        ])

        print("\nYour matrix:")
        print(A)

        img_array = np.array(image)
        h, w = img_array.shape[:2]

        cx = (w - 1) / 2
        cy = (h - 1) / 2

        X, Y = np.meshgrid(
            np.arange(w),
            np.arange(h)
        )

        Xc = X - cx
        Yc = Y - cy

        new_X = A[0, 0] * Xc + A[0, 1] * Yc
        new_Y = A[1, 0] * Xc + A[1, 1] * Yc

        new_X += cx
        new_Y += cy

        min_x = int(np.floor(new_X.min()))
        max_x = int(np.ceil(new_X.max()))
        min_y = int(np.floor(new_Y.min()))
        max_y = int(np.ceil(new_Y.max()))

        new_w = max_x - min_x + 1
        new_h = max_y - min_y + 1

        result = np.zeros(
            (new_h, new_w, 3),
            dtype=np.uint8
        )

        new_X = np.rint(new_X - min_x).astype(int)
        new_Y = np.rint(new_Y - min_y).astype(int)

        for y in range(h):
            for x in range(w):

                nx = new_X[y, x]
                ny = new_Y[y, x]

                if 0 <= nx < new_w and 0 <= ny < new_h:
                    result[ny, nx] = img_array[y, x]

        image = Image.fromarray(result)

        show_image(image, "Custom Matrix Transformation")


    # Reset
    elif choice == "6":
        image = original.copy()

        show_image(image, "Original Image")


    # Exit
    elif choice == "7":
        print("Toolbox closed.")
        break

    else:
        print("Invalid choice. Please try again.")
```
